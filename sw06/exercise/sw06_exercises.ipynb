## Exercise 2: Letters and digits

Write a python program that asks for a string as input and calculates the number of digits and letters contained.

Example:

```
Input = "python 3.2"
```

Expected result:

```
Letters: 6
Digits: 2
```

HINT: Check out the documentation for `string.isdigit()` and `string.isalpha()`.

Comment: Provide the solution method as part of a class.

````python
class StringAnalyzer:

    def __init__(self, text):
        self.text = text

    def count_letters_and_digits(self):
        letters = 0
        digits = 0
        for char in self.text:
            if char.isalpha():
                letters += 1
            elif char.isdigit():
                digits += 1
        return letters, digits


text = input("Enter a string: ")
analyzer = StringAnalyzer(text)
letters, digits = analyzer.count_letters_and_digits()
print(f"Letters: {letters}")
print(f"Digits: {digits}")
````
