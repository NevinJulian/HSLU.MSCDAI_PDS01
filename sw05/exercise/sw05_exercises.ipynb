## Read and write files

### Exercise 1: Combine three lines

Write a program that combines three lines from the file `musicians.txt` into one line and writes them to standard output using `print()`. Hence, an output line should contain first name, last name and artist name or band.

The band or artist name should be printed in round brackets and in capital letters.

Example: `Marshall Bruce Mathers III (EMINEM)`

````python
f = open("musicians.txt", "r")

counter = 0
new_line = ""

for line in f:
    line = line.rstrip()
    if counter % 3 == 0:
        new_line = line
    elif counter % 3 == 1:
        new_line += " " + line
    else:
        new_line += " (" + line.upper() + ")"
        print(new_line)
    counter += 1
    
f.close()
````

### Exercise 2: Write to a file

Change the solution of the previous exercise such that the program writes the output to a potentially newly created file `musicians2.txt`.

````python
f = open("musicians.txt", "r")
f2 = open("musicians2.txt", "w")

counter = 0
new_line = ""

for line in f:
    line = line.rstrip()
    if counter % 3 == 0:
        new_line = line
    elif counter % 3 == 1:
        new_line += " " + line
    else:
        new_line += " (" + line.upper() + ")"
        f2.write(new_line + "\n")
    counter += 1
    
f.close()
f2.close()
````
