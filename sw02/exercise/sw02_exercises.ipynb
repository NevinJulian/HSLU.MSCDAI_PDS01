## Exercise 1: Day of the week

Write a program that asks for a number between 1 and 7. The number stands for a day:
1 = Monday, 2 = Tuesday, ..., 7 = Sunday.

Print a message depending on the input:

- `You have to work ...!` for Monday to Friday
- `Enjoy your weekend...!` for Saturday and Sunday
- `Wrong input...!` for a wrong number

````python
daynumber = int(input("Input the day of Week: (1 = Montag, 2 = Dienstag, ..., 7 = Sonntag) "))

if daynumber == 1:
    print("Montag")
elif daynumber == 2:
    print("Dienstag")
elif daynumber == 3:
    print("Mittwoch")
elif daynumber == 4:
    print("Donnerstag")
elif daynumber == 5:
    print("Freitag")
elif daynumber == 6:
    print("Samstag")
elif daynumber == 7:
    print("Sonntag")
else:
    print("Ungültige Eingabe. Bitte geben Sie eine Zahl zwischen 1 und 7 ein.")
````

## Exercise 3: Temperature conversion

Convert temperatures between degrees Celsius and degrees Fahrenheit. Water freezes at 0 °C or 32 °F and boils at 100 °C or 212 °F. The two scales are related by

$$T_C = \frac{5}{9} \cdot (T_F - 32)$$

Write two programs:

- Celsius to Fahrenheit
- Fahrenheit to Celsius

Then think about how to combine both in one program, using an additional input for the direction, e.g. `F` and `45` for Fahrenheit to Celsius.

````python
temperatureMetric = str(input("Input the temperature metric: (C = Celsius, F = Fahrenheit) ")).upper()

if temperatureMetric == "C":
    temperature = float(input("Input the temperature in Celsius: "))
    fahrenheit = (temperature * 9/5) + 32
    print(f"{temperature}°C is equal to {fahrenheit}°F.")
elif temperatureMetric == "F":
    temperature = float(input("Input the temperature in Fahrenheit: "))
    celsius = (temperature - 32) * 5/9
    print(f"{temperature}°F is equal to {celsius}°C.")
else:
    print("Ungültige Eingabe. Bitte geben Sie 'C' für Celsius oder 'F' für Fahrenheit ein.")
````
